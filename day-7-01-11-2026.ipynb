GROUP ANAGRAMS 


```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        grp = {}
        
        for i in str :  
            label = "".join(sorted(i))
            if label not in grp :
                grp[label] = []
            grp[label].append(i)
        return list(grp.values())
```
