```python
def is_palindrome(s):
    return s == s[::-1 ]

word = "Tilak"
print(word[::-1])
print(word[::2])
print(word[1:4:1])
```

```python
#day end code 
def sum_list(nums) : 
    total = 0
    for x in nums :
        total = total + x
    return total    
        
mylist = [1,2,3]
print(sum_list(mylist))    
```
