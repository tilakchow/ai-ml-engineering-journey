```python
#Write has_duplicate(nums). It returns True if any number appears more than once.
def has_duplicate(num) : 
    seen = set()
    for x in num :
        if x in seen :
            print(seen)
            return True 
        else :
            seen.add(x)
    print(seen)        
    return False

num = [1,2,7,7,8,8,9]

print(has_duplicate(num))
                 
        
```

```python
def has_duplicate(nums) :
     seen = set()
     for x in nums :
        if x in seen :
            print(seen)
            return True
        else :
            seen.add(x)
     print(seen)
     return False

print(has_duplicate([1,2,3,3]))
```

```python
def count_letters(s):
    count = {}
    for ch in s:
        count[ch] = count.get(ch, 0) + 1
    return count

print(count_letters("aab"))
```

```python
def unique_count(nums) :
    return len(set(nums))


print(unique_count([1, 2, 2, 3, 3, 3]))
```

```python
def word_length(words) :
    result = {}
    for w in words :
        result[w] = len(w)
    return result       
```

```python
def first_repeat(nums):
    seen = set()
    for x in nums :
        if x in seen:
            return x
        seen.add(x)
    return None            
    
print(first_repeat([1,2,3,4,2,1]))    

```
