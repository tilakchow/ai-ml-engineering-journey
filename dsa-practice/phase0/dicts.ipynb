```python
#counter problem 
from collections import Counter
c = Counter(["apple","bat","bat","cat","apple","apple"])
print(c)
```

```python
d = ["apple","bat","bat","cat","apple","apple"]
dic = {}
for i in d :
    if i in dic :
        dic[i] = dic[i]+1
    else :
        dic[i] = 1
print(dic)        
```

```python
from collections import defaultdict
nums = [1,2,3,4,5,6,7,8]
groups = defaultdict(list)
for n in nums :
    groups[n%2].append(n)

print(groups)
```

```python
from collections import defaultdict
nums = [10, 15, 22, 33, 44, 51]
grp = defaultdict(list)
for n in nums :
    grp[n%2].append(n)
print(grp)
```

```python
from collections import Counter
words = ["apple","bat","bat","cat","apple","apple"]
c = Counter(words)
print(c)
```
