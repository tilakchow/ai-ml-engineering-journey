```python
#sorted
num = [5,2,8,1]
print(sorted(num))
print(sorted(num,reverse=True))

words = ["Banana","guva","berry","fig"]
print(sorted(words, key=len))
```

```python
#lambda
lambda x: x*2
print(sorted(words, key=lambda w: len(w)))
print(sorted(words, key=lambda w: w[-1]))
```

```python
#Print each with its index using enumerate. Try it.
names = ["x","y","z"]
for i,v in enumerate(names) :
    print(i,v)
```

```python
#Pair them up and print 1 one, 2 two, 3 three using zip.
a = [1,2,3]
b = ["one","two","three"]
for n,s in zip(a,b):
    print(n,s)
```
