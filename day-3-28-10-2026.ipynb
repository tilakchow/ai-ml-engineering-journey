```python
def factorial(n) :
    if n == 0 :
        return 1
    else :
        num = n * factorial(n-1)
        return num

print(factorial(3))
```

```python
def factorial(n) :
    if n == 0 :
        return 1
    return n*factorial(n-1)    
    
print(factorial(3))
```

```python
def sum_list(nums) :
    if len(nums)== 0:
        return 0
    return nums[0] + sum_list(nums[1:])  
print(sum_list([1,2,3]))        
```

```python
def fibonacci(n) :
    if n == 0 : return 0
    else :
        if n == 1 : return 1
        return fibonacci(n-1) + fibonacci(n-2)
print(fibonacci(30))        
```
