## 1. Contains Duplicate (LeetCode 217)

### Approach & Steps (in plain English):
1. Initialize an empty hash set `seen`.
2. Iterate through each number in `nums`.
3. If the number is already in `seen`, return `True` (duplicate found).
4. Otherwise, add the number to `seen`.
5. If the loop finishes without finding duplicates, return `False`.

**Complexity:**
- Time: O(n) where n is len(nums)
- Space: O(n) to store up to n elements in the set

```python
class Solution:
    def containsDuplicate(self, nums: list[int]) -> bool:
        seen = set()
        for n in nums:
            if n in seen:
                return True
            seen.add(n)
        return False

# Tests
sol = Solution()
assert sol.containsDuplicate([1, 2, 3, 1]) == True
assert sol.containsDuplicate([1, 2, 3, 4]) == False
assert sol.containsDuplicate([1, 1, 1, 3, 3, 4, 3, 2, 4, 2]) == True
print("All Contains Duplicate test cases passed!")
```

## 2. Valid Anagram (LeetCode 242)

### Approach & Steps (in plain English):
1. Check if lengths of `s` and `t` are equal. If not, return `False` immediately.
2. Build character counts for `s` using a dictionary.
3. Decrement character counts while iterating through `t`.
4. If a char in `t` is missing or count is 0, return `False`.
5. Return `True` once all characters match.

**Complexity:**
- Time: O(n) where n is len(s)
- Space: O(1) (at most 26 lowercase English letters)

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False
        
        count = {}
        for ch in s:
            count[ch] = count.get(ch, 0) + 1
            
        for ch in t:
            if ch not in count or count[ch] == 0:
                return False
            count[ch] -= 1
            
        return True

# Tests
sol = Solution()
assert sol.isAnagram("anagram", "nagaram") == True
assert sol.isAnagram("rat", "car") == False
print("All Valid Anagram test cases passed!")
```

## 3. Two Sum (LeetCode 1)

### Approach & Steps (in plain English):
1. Initialize a hash map `prev_map` mapping value -> index.
2. Iterate through `nums` with index `i` and value `num`.
3. Calculate the complement: `diff = target - num`.
4. If `diff` exists in `prev_map`, return `[prev_map[diff], i]`.
5. Otherwise, store current value and index: `prev_map[num] = i`.

**Complexity:**
- Time: O(n) single pass
- Space: O(n) hash map

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        prev_map = {}
        for i, num in enumerate(nums):
            diff = target - num
            if diff in prev_map:
                return [prev_map[diff], i]
            prev_map[num] = i
        return []

# Tests
sol = Solution()
assert sol.twoSum([2, 7, 11, 15], 9) == [0, 1]
assert sol.twoSum([3, 2, 4], 6) == [1, 2]
assert sol.twoSum([3, 3], 6) == [0, 1]
print("All Two Sum test cases passed!")
```
